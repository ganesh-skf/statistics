# 📊 Hypothesis Testing: t-test, ANOVA, Chi-square

Hypothesis testing is a fundamental statistical method used to decide whether there is enough evidence in a sample to infer that a certain condition holds for the entire population.

---

## 🧪 What is Hypothesis Testing?

Hypothesis testing allows us to make inferences about population parameters based on sample data. It is a way to test the validity of a claim or hypothesis about a population.

### 🔹 Null Hypothesis (H₀):
The default assumption — no effect, no difference, or status quo.

### 🔹 Alternative Hypothesis (H₁ or Ha):
The claim we are trying to test — that there **is** an effect or difference.

### 🔹 Significance Level (α):
The threshold for rejecting H₀ (commonly 0.05).

### 🔹 p-value:
The probability of obtaining test results at least as extreme as the results actually observed, under the assumption that the null hypothesis is correct.

---

## ✅ Decision Rule

- If **p-value ≤ α** → Reject the null hypothesis (evidence supports H₁)
---

## 📌 **1. t-test (Student's t-test)**

Used to compare the **means** of **two groups**.

### 🔹 Types:
- **One-sample t-test**: Compare sample mean to a known value
- **Independent two-sample t-test**: Compare means of two independent groups


## **📌 2. ANOVA (Analysis of Variance)**

Used to compare means across more than two groups.

**Types:**
- One-way ANOVA: One categorical independent variable



**Hypotheses:**
- H₀: All group means are equal

- H₁: At least one group mean is different

## **📌 3. Chi-square Test (χ²)**

Used for categorical variables to test the relationship between them.

**Types:**
- Chi-square Test of Independence: Are two categorical variables related?



## **✅ Conclusion**

Hypothesis testing is crucial for drawing conclusions from data:

- t-tests test mean differences between two groups

- ANOVA compares means across multiple groups

- Chi-square checks relationships between categorical variables

These tools enable data-driven decision-making in business, science, healthcare, and beyond.

## 🧪 1. Independent t-test
**Problem:**
You want to know whether there is a significant difference in average exam scores between male and female students in a class.

### 🔍 Hypotheses:
- **H₀ (Null Hypothesis)**: μ_male = μ_female (no difference in means)
- **H₁ (Alternative Hypothesis)**: μ_male ≠ μ_female (means are different)

```python
import numpy as np
from scipy.stats import ttest_ind

# Generate synthetic data
np.random.seed(1)
male_scores = np.random.normal(loc=70, scale=10, size=30)
female_scores = np.random.normal(loc=75, scale=12, size=30)

# Perform independent t-test
t_stat, p_value = ttest_ind(male_scores, female_scores)

# Print results
print(f"t-statistic = {t_stat:.2f}, p-value = {p_value:.4f}")
```

## 🧪 2. One-Way ANOVA
**Problem:**
You want to check whether students from three different colleges have different average GPAs.

## 🔍 Hypotheses:
- H₀: μ_A = μ_B = μ_C (all group means are equal)

- H₁: At least one group mean is different

```python
from scipy.stats import f_oneway
import numpy as np

# Simulate GPAs from 3 colleges
# np.random.seed(2)
college_A = np.random.normal(3.0, 0.29, 30)
college_B = np.random.normal(3.2, 0.25, 30)
college_C = np.random.normal(3.1, 0.4, 30)

# Perform one-way ANOVA
f_stat, p_value = f_oneway(college_A, college_B, college_C)

print(f"F-statistic = {f_stat:.2f}, p-value = {p_value:.4f}")
```

```python
college_C
```

## **🧪 3. Chi-square Test of Independence**
**Problem:**
You want to know whether gender is associated with preferred mobile plan type (Basic, Standard, Premium).

## 🔍 Hypotheses:
- H₀: Gender and plan preference are independent

- H₁: Gender and plan preference are dependent (associated)



```python
import pandas as pd
from scipy.stats import chi2_contingency

# Create contingency table
data = {
    'Basic': [20, 30],       # [Males, Females]
    'Standard': [25, 20],
    'Premium': [15, 20]
}
contingency_table = pd.DataFrame(data, index=['Male', 'Female'])

# Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square = {chi2:.2f}, p-value = {p:.4f}")
```
