```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from statistics import mode

print("Libraries Imported Successfully")
```


# 1. Mean, Median and Mode


```python
data=np.random.random_integers(1,1000,100)
```

```python
data
```

```python

# data = [10, 20, 20, 30, 40, 50]

mean_value = np.mean(data)
median_value = np.median(data)
mode_value = mode(data)

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```


# 2. Variance and Standard Deviation


```python

data = [2, 4, 6, 8, 10]

variance = np.var(data)
std_deviation = np.std(data)

print("Variance:", variance)
print("Standard Deviation:", std_deviation)
```


# 3. Distribution Visualization


```python
sns.histplot(data)
```

```python

# Generate normal distribution data
normal_data = np.random.normal(loc=100, scale=10, size=1000)

# plt.figure(figsize=(8,5))
sns.histplot(normal_data, kde=True)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
mean_value = np.mean(normal_data)
median_value = np.median(normal_data)
mode_value = mode(normal_data)

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```


# 4. Skewness and Kurtosis


```python

skewness = stats.skew(normal_data)
kurtosis = stats.kurtosis(normal_data)

print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
```


# 5. Z-Score Calculation


```python

sample_data = [45, 50, 55, 60, 65]

z_scores = stats.zscore(sample_data)

print("Z-Scores:")
print(z_scores)
```

```python
sns.boxplot(normal_data)
```


# 6. Detecting Outliers using Z-Score


```python

# data = [10, 12, 14, 15, 18]

z = np.abs(stats.zscore(data))

print("Z-Scores:",z)
outliers = np.where(z>2)
print(outliers)
print("Outlier Index:", np.array(data)[outliers])
```

```python
# from scipy import stats

# data = [10, 12, 14, 15, 18, 100]

# Calculate Z-scores
z = np.abs(stats.zscore(data))

print("Z-Scores:", z)

# Find outliers using threshold 2
outliers = np.where(z > 2)

print("Outlier Index:", outliers)
print("Outlier Value:", np.array(data)[outliers])
```


# 7. Covariance and Correlation


```python

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

covariance = np.cov(x, y)[0][1]
correlation = np.corrcoef(x, y)[0][1]

print("Covariance:", covariance)
print("Correlation:", correlation)
```


# 8. Correlation Heatmap


```python

df = pd.DataFrame({
    'Maths': [70, 80, 90, 60, 50],
    'Science': [72, 85, 88, 65, 55],
    'English': [68, 78, 92, 58, 48]
})

corr = df.corr()

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```


# 9. Q-Q Plot


```python

stats.probplot(normal_data, dist="norm", plot=plt)
plt.title("Q-Q Plot")
plt.show()
```


# 10. Boxplot for Outlier Detection


```python

data = [10, 12, 13, 15, 16, 18,10,100]

plt.figure(figsize=(6,4))
sns.boxplot(data=data)

plt.title("Boxplot for Outlier Detection")

plt.show()
```


# 11. Distribution Transformation


```python
import numpy as np
```

```python

skewed_data = np.random.exponential(scale=2, size=1000)

log_transformed = np.log1p(skewed_data)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
sns.histplot(skewed_data, kde=True)
plt.title("Original Skewed Data")

plt.subplot(1,2,2)
sns.histplot(log_transformed, kde=True)
plt.title("Log Transformed Data")

plt.show()
```
